## Spill Suksess Prediktor

I dette prosjektet, skal vi prøve å forutsi om et steam spill kommer til å vellykket før det blir gitt ut.
Vi skal se på datasettet, trene modellen og deretter avluere.

## 1.1 Datasettet

Vi bruker pandas for å loade Steam-spill-datasettet fra en CSV-fil.

```python
import pandas as pd

data = pd.read_csv("games.csv")

data.head()
```

Vi leser de første 5 radene og ser at verdiene ligger under riktig kolonne.

## 1.2 Undersøke størrelsen på datasettet

Vi undersøker hvor mange spill og kolonner datasettet har. Det gir oss en oversikt over hvor mange spill vi har tilgjengelig og hvor mange egenskaper for hvert spill.

```python
print("Antall spill:", data.shape[0])
print("Antall kolonner:", data.shape[1])
```

Datasettet inneholder 125855 spill og 40 kolonner.

## 1.3 Undersøke estimert antall eiere

Vi må undersøke fordelingen av estimert antall eiere for å forstå hvor populære spillene i datasettet er. Dette vil hjelpe oss med å definere hva som kan være et "vellykket" spill.

```python
data["Estimated owners"].value_counts()
```

## 1.3.1 Observasjoner

Vi ser at datasettet inneholder flest spill med lavt estimert antall eiere.

Dersom vi definerer et populært spill som et spill med minst 100 000 estimerte eiere, vil 8 819 av 125 855 spill (ca. 7 %) bli klassifisert som populære. 

Dette gir en ubalansert fordeling mellom de to klassene, noe vi må ta hensyn til når vi trener og evaluerer modellen.

## 1.4 Opprette target-variablen

Vi ønsker å utvikle en maskinlæringsmodell som kan forutsi om et spill blir populært.

Vi definerer foreløpig et populært spill som et spill med minst 100 000 estimerte eiere.

For å gjøre dette til et klassifikasjonsproblem oppretter vi en ny kolonne kalt Successful, der:

0 betyr ikke populært.

1 betyr populært.

Denne kolonnen skal senere brukes som modellens target (y), altså verdien modellen skal lære å forutsi.
